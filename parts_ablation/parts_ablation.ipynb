```python
import seaborn as sns
import matplotlib.pyplot as plt

model_rename = {
    "tabm_empty": "MLP",
    "tabm": "Emb+SurvHL+Ens",
    "tabm_non_ens": "SurvHL+Ens",
    "tabm_non_loss": "Emb+Ens",
    "tabm_non_rtdl": "Emb+SurvHL",
    "tabm_only_ens": "Ens",
    "tabm_only_loss": "SurvHL",
    "tabm_only_rtdl": "Emb",
}

cidx = tables["c_index_relative"].copy()
cidx = cidx.drop(index="MEAN", errors="ignore")
cidx = cidx.rename(columns=model_rename)

order = cidx.mean(axis=0).sort_values(ascending=True).index.tolist()

plot_df = (
    cidx[order]
    .reset_index()
    .melt(id_vars="dataset", var_name="model", value_name="relative_c_index")
)

plt.figure(figsize=(6, 8))
sns.boxplot(
    data=plot_df,
    y="model",
    x="relative_c_index",
    order=order,
    hue="model",
    showfliers=False,
)

# sns.stripplot(
#     data=plot_df,
#     x="model",
#     y="relative_c_index",
#     order=order,
#     hue="dataset",
#     dodge=False,
#     alpha=0.75,
#     size=5,
# )

# plt.axhline(0, color="black", linestyle="--", linewidth=1)
plt.ylabel("")
plt.xlabel("Relative C-index, %")
plt.title("C-index relative improvement by model")
# plt.xticks(rotation=45, ha="right")
# plt.legend(title="dataset", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
ibs = tables["ibs_relative"].copy()
ibs = ibs.drop(index="MEAN", errors="ignore")
ibs = ibs.rename(columns=model_rename)

order = ibs.mean(axis=0).sort_values(ascending=True).index.tolist()

plot_df = (
    ibs[order]
    .reset_index()
    .melt(id_vars="dataset", var_name="model", value_name="relative_ibs")
)

plt.figure(figsize=(6, 8))
sns.boxplot(
    data=plot_df,
    y="model",
    x="relative_ibs",
    order=order,
    hue="model",
    showfliers=False,
)

plt.ylabel("")
plt.xlabel("Relative IBS, %")
plt.title("IBS relative improvement by model")
plt.tight_layout()
plt.show()
```

```python
auc = tables["auc_relative"].copy()
auc = auc.drop(index="MEAN", errors="ignore")
auc = auc.rename(columns=model_rename)

order = auc.mean(axis=0).sort_values(ascending=True).index.tolist()

plot_df = (
    auc[order]
    .reset_index()
    .melt(id_vars="dataset", var_name="model", value_name="relative_auc")
)

plt.figure(figsize=(6, 8))
sns.boxplot(
    data=plot_df,
    y="model",
    x="relative_auc",
    order=order,
    hue="model",
    showfliers=False,
)

plt.ylabel("")
plt.xlabel("Relative AUC, %")
plt.title("AUC relative improvement by model")
plt.tight_layout()
plt.show()
```
